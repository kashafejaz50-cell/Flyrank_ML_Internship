## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*


My lane is a **ranking** task. The goal is to rank webpages based on how urgently they should be reviewed for a content refresh. This helps content editors prioritize their work by focusing on the pages most likely to benefit from updates. Ranking is the most suitable task because the decision is not simply whether a page needs a refresh, but **which pages should be reviewed first**.

```python
# This cell is for CODE (numbers, a query, a check).
ml_task = "Ranking"

print("ML Task Type:", ml_task)
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*


The target is the observed content performance trend of each webpage, which is used to estimate its refresh priority. Rather than creating a label using a manual rule (for example, "refresh pages with low clicks"), the model should learn from observed search and engagement outcomes. The predicted output is a ranking that helps content editors identify which webpages should be reviewed first.

```python
# This cell is for CODE (numbers, a query, a check).
signal_columns = [
    "impressions_90d",
    "clicks_90d",
    "ctr",
    "avg_position",
    "engagement_rate"
]

print("Signals that could help estimate refresh priority:")
print(df[signal_columns].head())
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*


The success metric is **Precision@10**. This measures how many of the top 10 webpages recommended by the model actually deserve priority for a content refresh. A good model should place the most valuable pages near the top of the ranked list so that content editors can focus their efforts on the highest-impact updates.

```python
# This cell is for CODE (numbers, a query, a check).
success_metric = "Precision@10"

print("Success Metric:", success_metric)
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule cannot reliably decide which webpages should be refreshed because content performance depends on many interacting factors. Search volume, impressions, clicks, CTR, average position, engagement rate, content age, and freshness all influence whether a page should be reviewed. These relationships are complex and can vary across different webpages and clients. A machine learning model can learn patterns from multiple signals simultaneously and produce better ranking recommendations than a simple if-statement. The model is intended to support editorial decisions, not replace them.

```python
# This cell is for CODE (numbers, a query, a check).
# Features that could help describe content performance
signals = [
    "search_volume",
    "impressions_90d",
    "clicks_90d",
    "ctr",
    "avg_position",
    "engagement_rate",
    "content_age_days",
    "days_since_last_update"
]

df[signals].describe()
```
