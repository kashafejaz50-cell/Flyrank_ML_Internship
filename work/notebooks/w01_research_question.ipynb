## 3. Quick look at the data (2-3 real numbers)

## Quick Look at the Data
The starter dataset contains 30,000 pages and 44 columns, providing enough data for exploratory analysis.

The correlation between search volume and 90-day impressions is approximately 0.001, indicating that higher search volume alone does not necessarily lead to more impressions.

The median word count is similar for both declining and growing pages, suggesting that content length alone is not the main factor behind content performance.

These observations indicate that effective content review decisions require multiple observable signals rather than relying on a single metric.

```python
import pandas as pd

# Load the starter dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

# Discovery A
corr = df["search_volume"].corr(df["impressions_90d"])
print("Correlation between search volume and impressions:", round(corr, 3))

# Discovery C
median_wc = df.groupby("trend_direction")["word_count"].median()

print("\nMedian word count by trend:")
print(median_wc)
```
